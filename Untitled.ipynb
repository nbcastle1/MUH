```python
import os
import glob
from PyPDF2 import PdfWriter, PdfReader
from pathlib import Path
import re

def find_pdfs_with_keyword(root_directory, keywords, search_in="filename", case_sensitive=False):
    """
    Find PDF files containing keywords in filename or content
    
    Args:
        root_directory (str): Root directory to search in
        keywords (list or str): Keywords to search for
        search_in (str): "filename", "content", or "both"
        case_sensitive (bool): Whether search should be case sensitive
    
    Returns:
        list: List of matching PDF file paths
    """
    
    if isinstance(keywords, str):
        keywords = [keywords]
    
    # Find all PDF files recursively
    pdf_files = []
    for root, dirs, files in os.walk(root_directory):
        for file in files:
            if file.lower().endswith('.pdf'):
                pdf_files.append(os.path.join(root, file))
    
    print(f"Found {len(pdf_files)} PDF files total in directory tree...")
    
    matching_files = []
    
    for pdf_file in pdf_files:
        filename = os.path.basename(pdf_file)
        matches_filename = False
        matches_content = False
        
        # Check filename for keywords
        if search_in in ["filename", "both"]:
            search_text = filename if case_sensitive else filename.lower()
            search_keywords = keywords if case_sensitive else [k.lower() for k in keywords]
            
            matches_filename = any(keyword in search_text for keyword in search_keywords)
        
        # Check content for keywords (if requested)
        if search_in in ["content", "both"]:
            try:
                with open(pdf_file, 'rb') as file:
                    pdf_reader = PdfReader(file)
                    content = ""
                    
                    # Extract text from first few pages for performance
                    max_pages = min(5, len(pdf_reader.pages))
                    for page_num in range(max_pages):
                        try:
                            page_text = pdf_reader.pages[page_num].extract_text()
                            content += page_text + " "
                        except:
                            continue
                    
                    search_text = content if case_sensitive else content.lower()
                    search_keywords = keywords if case_sensitive else [k.lower() for k in keywords]
                    
                    matches_content = any(keyword in search_text for keyword in search_keywords)
                    
            except Exception as e:
                print(f"Warning: Could not read content from {filename}: {str(e)}")
                matches_content = False
        
        # Add file if it matches criteria
        if (search_in == "filename" and matches_filename) or \
           (search_in == "content" and matches_content) or \
           (search_in == "both" and (matches_filename or matches_content)):
            matching_files.append(pdf_file)
            print(f"✓ Match: {filename}")
    
    print(f"\nFound {len(matching_files)} PDFs matching keyword criteria")
    return matching_files

def combine_pdfs_by_keyword(root_directory, keywords, output_filename, 
                          search_in="filename", case_sensitive=False, 
                          add_bookmarks=True, sort_files=True):
    """
    Combine all PDFs containing keywords from directory and subdirectories
    
    Args:
        root_directory (str): Root directory to search
        keywords (list or str): Keywords to search for
        output_filename (str): Output PDF filename
        search_in (str): "filename", "content", or "both"
        case_sensitive (bool): Case sensitive search
        add_bookmarks (bool): Add bookmarks for each original file
        sort_files (bool): Sort files before combining
    """
    
    # Find matching PDF files
    matching_files = find_pdfs_with_keyword(
        root_directory, keywords, search_in, case_sensitive
    )
    
    if not matching_files:
        print("No matching PDF files found!")
        return
    
    if sort_files:
        matching_files.sort()
    
    # Create PDF writer
    pdf_writer = PdfWriter()
    current_page = 0
    successful_files = 0
    
    print(f"\nCombining {len(matching_files)} PDFs...")
    
    for i, pdf_file in enumerate(matching_files):
        try:
            filename = os.path.basename(pdf_file)
            relative_path = os.path.relpath(pdf_file, root_directory)
            
            print(f"Processing {i+1}/{len(matching_files)}: {relative_path}")
            
            with open(pdf_file, 'rb') as file:
                pdf_reader = PdfReader(file)
                
                # Add bookmark if requested
                if add_bookmarks:
                    bookmark_title = f"{Path(pdf_file).stem} ({os.path.dirname(relative_path)})"
                    pdf_writer.add_outline_item(bookmark_title, current_page)
                
                # Add all pages
                for page in pdf_reader.pages:
                    pdf_writer.add_page(page)
                    current_page += 1
                
                successful_files += 1
                
        except Exception as e:
            print(f"Error processing {pdf_file}: {str(e)}")
            continue
    
    # Write combined PDF
    try:
        with open(output_filename, 'wb') as output_file:
            pdf_writer.write(output_file)
        
        print(f"\n✅ Successfully combined {successful_files} PDFs into {output_filename}")
        print(f"Total pages: {len(pdf_writer.pages)}")
        
        # Show file size
        file_size = os.path.getsize(output_filename) / (1024 * 1024)  # MB
        print(f"File size: {file_size:.1f} MB")
        
    except Exception as e:
        print(f"❌ Error writing output file: {str(e)}")

def combine_with_regex_pattern(root_directory, pattern, output_filename):
    """
    Combine PDFs matching a regex pattern in filename
    
    Args:
        root_directory (str): Root directory to search
        pattern (str): Regex pattern to match filenames
        output_filename (str): Output filename
    """
    
    pdf_files = []
    regex = re.compile(pattern, re.IGNORECASE)
    
    for root, dirs, files in os.walk(root_directory):
        for file in files:
            if file.lower().endswith('.pdf') and regex.search(file):
                pdf_files.append(os.path.join(root, file))
                print(f"✓ Pattern match: {file}")
    
    if not pdf_files:
        print(f"No files matching pattern '{pattern}' found!")
        return
    
    pdf_files.sort()
    pdf_writer = PdfWriter()
    
    for pdf_file in pdf_files:
        try:
            with open(pdf_file, 'rb') as file:
                pdf_reader = PdfReader(file)
                for page in pdf_reader.pages:
                    pdf_writer.add_page(page)
        except Exception as e:
            print(f"Error processing {pdf_file}: {str(e)}")
    
    with open(output_filename, 'wb') as output_file:
        pdf_writer.write(output_file)
    
    print(f"Combined {len(pdf_files)} PDFs matching pattern into {output_filename}")

# Usage Examples
if __name__ == "__main__":
    
    # Example 1: Search by filename keywords
    root_dir = "C:/Users/castle/Downloads/ConsentDocs (1)/ConsentDocs"
    keywords = ["Data", "Collection", "Sheet"]  # Files containing any of these words
    output_file = "C:/Users/castle/Downloads/combined_reports.pdf"
    
    combine_pdfs_by_keyword(
        root_directory=root_dir,
        keywords=keywords,
        output_filename=output_file,
        search_in="filename",  # Search in filename only
        case_sensitive=False,
        add_bookmarks=True
    )
```

```python
import cv2
import numpy as np
import pandas as pd
import pytesseract
from PIL import Image
import fitz  # PyMuPDF
import re
import json
from typing import List, Dict, Tuple
import matplotlib.pyplot as plt
import seaborn as sns
from dataclasses import dataclass
from pathlib import Path

@dataclass
class OCRResult:
    text: str
    confidence: float
    bbox: Tuple[int, int, int, int]  # x, y, width, height
    page_num: int

class PDFOCRTableExtractor:
    def __init__(self, tesseract_path=None):
        """
        Initialize the PDF OCR Table Extractor
        
        Args:
            tesseract_path (str): Path to tesseract executable (if not in PATH)
        """
        if tesseract_path:
            pytesseract.pytesseract.tesseract_cmd = tesseract_path
        
        # OCR configuration for better accuracy
        self.ocr_config = r'--oem 3 --psm 6 -c tessedit_char_whitelist=0123456789ABCDEFGHIJKLMNOPQRSTUVWXYZabcdefghijklmnopqrstuvwxyz.,:-_()[]{}/$%@#&*+=<>?!;"\' '
        
    def pdf_to_images(self, pdf_path: str, dpi: int = 300) -> List[np.ndarray]:
        """Convert PDF pages to high-resolution images"""
        doc = fitz.open(pdf_path)
        images = []
        
        for page_num in range(doc.page_count):
            page = doc[page_num]
            # High DPI for better OCR accuracy
            mat = fitz.Matrix(dpi/72, dpi/72)
            pix = page.get_pixmap(matrix=mat)
            img_data = pix.tobytes("ppm")
            
            # Convert to OpenCV format
            nparr = np.frombuffer(img_data, np.uint8)
            img = cv2.imdecode(nparr, cv2.IMREAD_COLOR)
            images.append(img)
            
        doc.close()
        return images
    
    def preprocess_image(self, image: np.ndarray) -> np.ndarray:
        """Preprocess image for better OCR results"""
        # Convert to grayscale
        gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
        
        # Apply Gaussian blur to reduce noise
        blurred = cv2.GaussianBlur(gray, (1, 1), 0)
        
        # Apply threshold to get binary image
        _, thresh = cv2.threshold(blurred, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
        
        # Morphological operations to clean up
        kernel = np.ones((1, 1), np.uint8)
        cleaned = cv2.morphologyEx(thresh, cv2.MORPH_CLOSE, kernel)
        
        return cleaned
    
    def extract_text_with_confidence(self, image: np.ndarray, page_num: int) -> List[OCRResult]:
        """Extract text with confidence scores from image"""
        processed_img = self.preprocess_image(image)
        
        # Get detailed OCR data
        data = pytesseract.image_to_data(processed_img, config=self.ocr_config, output_type=pytesseract.Output.DICT)
        
        ocr_results = []
        n_boxes = len(data['level'])
        
        for i in range(n_boxes):
            confidence = float(data['conf'][i])
            text = data['text'][i].strip()
            
            # Filter out low confidence and empty results
            if confidence > 30 and text:  # Confidence threshold
                x, y, w, h = data['left'][i], data['top'][i], data['width'][i], data['height'][i]
                
                ocr_results.append(OCRResult(
                    text=text,
                    confidence=confidence,
                    bbox=(x, y, w, h),
                    page_num=page_num
                ))
        
        return ocr_results
    
    def detect_table_structure(self, ocr_results: List[OCRResult]) -> Dict:
        """Detect potential table structure from OCR results"""
        if not ocr_results:
            return {"rows": [], "confidence_stats": {}}
        
        # Group text by approximate Y coordinates (rows)
        y_coords = [result.bbox[1] for result in ocr_results]
        y_tolerance = 10  # pixels
        
        # Cluster Y coordinates to find rows
        y_sorted = sorted(set(y_coords))
        row_groups = []
        current_group = [y_sorted[0]]
        
        for y in y_sorted[1:]:
            if y - current_group[-1] <= y_tolerance:
                current_group.append(y)
            else:
                row_groups.append(current_group)
                current_group = [y]
        row_groups.append(current_group)
        
        # Create rows by grouping OCR results
        rows = []
        for group in row_groups:
            row_results = []
            for result in ocr_results:
                if result.bbox[1] in group:
                    row_results.append(result)
            
            if row_results:
                # Sort by X coordinate within row
                row_results.sort(key=lambda x: x.bbox[0])
                rows.append(row_results)
        
        # Calculate confidence statistics
        all_confidences = [result.confidence for result in ocr_results]
        confidence_stats = {
            "mean_confidence": np.mean(all_confidences),
            "min_confidence": np.min(all_confidences),
            "max_confidence": np.max(all_confidences),
            "std_confidence": np.std(all_confidences),
            "total_words": len(ocr_results)
        }
        
        return {
            "rows": rows,
            "confidence_stats": confidence_stats
        }
    
    def create_structured_table(self, table_data: Dict) -> pd.DataFrame:
        """Create a structured pandas DataFrame from detected table structure"""
        rows = table_data["rows"]
        
        if not rows:
            return pd.DataFrame()
        
        # Find the maximum number of columns
        max_cols = max(len(row) for row in rows) if rows else 0
        
        # Create table data
        table_rows = []
        confidence_rows = []
        
        for row in rows:
            # Extract text and confidence
            row_text = []
            row_conf = []
            
            for result in row:
                row_text.append(result.text)
                row_conf.append(result.confidence)
            
            # Pad with empty strings if needed
            while len(row_text) < max_cols:
                row_text.append("")
                row_conf.append(0.0)
            
            table_rows.append(row_text)
            confidence_rows.append(row_conf)
        
        # Create DataFrame
        df = pd.DataFrame(table_rows, columns=[f'Column_{i+1}' for i in range(max_cols)])
        
        # Add confidence information
        conf_df = pd.DataFrame(confidence_rows, columns=[f'Confidence_Col_{i+1}' for i in range(max_cols)])
        
        return df, conf_df
    
    def process_pdf(self, pdf_path: str, output_dir: str = "output") -> Dict:
        """Main function to process PDF and extract table with confidence scores"""
        
        # Create output directory
        Path(output_dir).mkdir(exist_ok=True)
        
        print(f"Processing PDF: {pdf_path}")
        
        # Convert PDF to images
        print("Converting PDF to images...")
        images = self.pdf_to_images(pdf_path)
        print(f"Extracted {len(images)} pages")
        
        # Process each page
        all_ocr_results = []
        page_results = {}
        
        for page_num, image in enumerate(images):
            print(f"Processing page {page_num + 1}...")
            ocr_results = self.extract_text_with_confidence(image, page_num)
            all_ocr_results.extend(ocr_results)
            page_results[f'page_{page_num + 1}'] = ocr_results
            
            # Save processed image for reference
            processed_img = self.preprocess_image(image)
            cv2.imwrite(f"{output_dir}/processed_page_{page_num + 1}.png", processed_img)
        
        print(f"Extracted {len(all_ocr_results)} text elements total")
        
        # Detect table structure
        print("Detecting table structure...")
        table_data = self.detect_table_structure(all_ocr_results)
        
        # Create structured table
        if table_data["rows"]:
            df, conf_df = self.create_structured_table(table_data)
            
            # Save to CSV
            df.to_csv(f"{output_dir}/extracted_table.csv", index=False)
            conf_df.to_csv(f"{output_dir}/confidence_scores.csv", index=False)
            
            # Create combined table with confidence
            combined_data = []
            for i in range(len(df)):
                row_data = {}
                for col in df.columns:
                    conf_col = f"Confidence_{col}"
                    row_data[col] = df.iloc[i][col]
                    if conf_col.replace("Confidence_", "Confidence_Col_") in conf_df.columns:
                        row_data[conf_col] = conf_df.iloc[i][conf_col.replace("Confidence_", "Confidence_Col_")]
                combined_data.append(row_data)
            
            combined_df = pd.DataFrame(combined_data)
            combined_df.to_csv(f"{output_dir}/table_with_confidence.csv", index=False)
            
        else:
            df = pd.DataFrame()
            conf_df = pd.DataFrame()
            combined_df = pd.DataFrame()
        
        # Create confidence visualization
        self.create_confidence_report(table_data["confidence_stats"], all_ocr_results, output_dir)
        
        # Save detailed OCR results
        ocr_export = []
        for result in all_ocr_results:
            ocr_export.append({
                'text': result.text,
                'confidence': result.confidence,
                'x': result.bbox[0],
                'y': result.bbox[1],
                'width': result.bbox[2],
                'height': result.bbox[3],
                'page': result.page_num + 1
            })
        
        with open(f"{output_dir}/ocr_results.json", 'w') as f:
            json.dump(ocr_export, f, indent=2)
        
        results = {
            "extracted_table": df,
            "confidence_scores": conf_df,
            "combined_table": combined_df,
            "ocr_results": all_ocr_results,
            "confidence_stats": table_data["confidence_stats"],
            "page_results": page_results
        }
        
        print(f"\nProcessing complete! Results saved to '{output_dir}' directory")
        print(f"Overall OCR confidence: {table_data['confidence_stats'].get('mean_confidence', 0):.1f}%")
        
        return results
    
    def create_confidence_report(self, stats: Dict, ocr_results: List[OCRResult], output_dir: str):
        """Create visual confidence report"""
        
        if not ocr_results:
            print("No OCR results to create confidence report")
            return
            
        confidences = [result.confidence for result in ocr_results]
        
        # Create figure with subplots
        fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(15, 10))
        
        # Confidence distribution histogram
        ax1.hist(confidences, bins=20, alpha=0.7, color='skyblue', edgecolor='black')
        ax1.set_xlabel('Confidence Score')
        ax1.set_ylabel('Frequency')
        ax1.set_title('OCR Confidence Distribution')
        ax1.axvline(stats['mean_confidence'], color='red', linestyle='--', 
                   label=f'Mean: {stats["mean_confidence"]:.1f}%')
        ax1.legend()
        
        # Confidence by page
        page_confidences = {}
        for result in ocr_results:
            page = result.page_num + 1
            if page not in page_confidences:
                page_confidences[page] = []
            page_confidences[page].append(result.confidence)
        
        pages = list(page_confidences.keys())
        mean_conf_by_page = [np.mean(page_confidences[p]) for p in pages]
        
        ax2.bar(pages, mean_conf_by_page, alpha=0.7, color='lightgreen')
        ax2.set_xlabel('Page Number')
        ax2.set_ylabel('Mean Confidence')
        ax2.set_title('Average Confidence by Page')
        
        # Confidence categories
        high_conf = len([c for c in confidences if c >= 80])
        med_conf = len([c for c in confidences if 60 <= c < 80])
        low_conf = len([c for c in confidences if c < 60])
        
        categories = ['High (≥80%)', 'Medium (60-79%)', 'Low (<60%)']
        counts = [high_conf, med_conf, low_conf]
        colors = ['green', 'orange', 'red']
        
        ax3.pie(counts, labels=categories, colors=colors, autopct='%1.1f%%', startangle=90)
        ax3.set_title('Confidence Categories')
        
        # Statistics table
        stats_text = f"""
        Total Words Detected: {stats['total_words']}
        Mean Confidence: {stats['mean_confidence']:.1f}%
        Min Confidence: {stats['min_confidence']:.1f}%
        Max Confidence: {stats['max_confidence']:.1f}%
        Std Deviation: {stats['std_confidence']:.1f}%
        
        Quality Assessment:
        {'Excellent' if stats['mean_confidence'] >= 90 else
         'Good' if stats['mean_confidence'] >= 75 else
         'Fair' if stats['mean_confidence'] >= 60 else
         'Poor'}
        """
        
        ax4.text(0.1, 0.5, stats_text, transform=ax4.transAxes, fontsize=11,
                verticalalignment='center', bbox=dict(boxstyle="round,pad=0.3", facecolor="lightblue"))
        ax4.set_xlim(0, 1)
        ax4.set_ylim(0, 1)
        ax4.axis('off')
        ax4.set_title('OCR Statistics Summary')
        
        plt.tight_layout()
        plt.savefig(f'{output_dir}/confidence_report.png', dpi=300, bbox_inches='tight')
        plt.close()
        
        print(f"Confidence report saved to {output_dir}/confidence_report.png")

# Usage example
def main():
    # Initialize the extractor
    extractor = PDFOCRTableExtractor()
    
    # Process the PDF
    pdf_path = "C:/Users/castle/Downloads/combined_reports.pdf"  # Your PDF file
    results = extractor.process_pdf(pdf_path, output_dir="C:/Users/castle/Downloads/ocr_output")
    
    # Display results summary
    if not results["extracted_table"].empty:
        print("\nExtracted Table Preview:")
        print(results["extracted_table"].head())
        print(f"\nTable shape: {results['extracted_table'].shape}")
        
        print("\nConfidence Summary:")
        print(f"Mean confidence: {results['confidence_stats']['mean_confidence']:.1f}%")
        print(f"Total words extracted: {results['confidence_stats']['total_words']}")
    else:
        print("No tabular data detected in the PDF")
    
    return results

if __name__ == "__main__":
    # Install required packages:
    # pip install opencv-python pandas pytesseract pillow PyMuPDF matplotlib seaborn numpy
    
    # Note: You also need to install Tesseract OCR:
    # Windows: Download from https://github.com/UB-Mannheim/tesseract/wiki
    # Mac: brew install tesseract
    # Linux: sudo apt-get install tesseract-ocr
    
    results = main()
```
